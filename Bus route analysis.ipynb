```python
import pandas as pd
df = pd.read_csv("college_bus_dataset12.csv")
stop_students = df.groupby('stop_name')['students'].mean()
print(stop_students)
```

```python
print("=== STOP-BASED IMPROVEMENTS ===\n")

avg = stop_students.mean()

for stop in stop_students.index:
    
    students = stop_students[stop]
    
    if students > avg + 1:
        print(f"{stop}: Overloaded ({students:.1f}) → Increase pickup frequency")
    
    elif students < avg - 1:
        print(f"{stop}: Underutilized ({students:.1f}) → Merge with nearby stop")
    
    else:
        print(f"{stop}: Balanced ({students:.1f}) → No change needed")
```
